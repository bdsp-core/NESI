```python
import mne
import os

def inspect_edf(edf_path):
    raw=mne.io.read_raw_edf(edf_path,preload=False,verbose=False)
    print(f"\nFile: {edf_path}")
    print(f"Filename: {os.path.basename(edf_path)}")
    print(f"\n--- RAW INFO KEYS ---")
    print(list(raw.info.keys()))
    print(f"\n--- BASIC INFO ---")
    print(f"Number of channels: {len(raw.ch_names)}")
    print(f"Sampling frequency: {raw.info['sfreq']} Hz")
    print(f"Number of samples: {raw.n_times}")
    print(f"Duration: {raw.n_times/raw.info['sfreq']:.2f} sec ({raw.n_times/raw.info['sfreq']/60:.2f} min)")
    print(f"Start time: {raw.info['meas_date']}")
    print(f"\n--- CHANNELS ---")
    for i,ch in enumerate(raw.ch_names):
        print(f"{i:3d}: {ch}")
    print(f"\n--- EDF/MNE INFO ---")
    for key,value in raw.info.items():
        print(f"{key}: {value}")
    print(f"\n--- ANNOTATIONS ---")
    print(f"Number of annotations: {len(raw.annotations)}")
    for i,(onset,duration,description) in enumerate(zip(raw.annotations.onset,raw.annotations.duration,raw.annotations.description)):
        print(f"{i}: onset={onset}, duration={duration}, description={description}")
    return raw

edf_path="/home/ayush/Desktop/NESI_review_material/ContiniousNESI_RASS_EEG/sub-S0001111365774_ses-1.edf"
raw=inspect_edf(edf_path)
```

```python
import os
import glob
import mat73
import pandas as pd

folder = "/home/ayush/Desktop/CAM-S_dataset/CAMS_EEG_modified"
mat_files = sorted(glob.glob(os.path.join(folder, "*.mat")))
rows = []

for i, file in enumerate(mat_files, 1):
    data = mat73.loadmat(file, only_include=["CAM_S_SF", "EEG_SessionID", "EvalDTS", "SourceEEG_StartDTS"])
    rows.append({
        "MorgothOutputFilename": os.path.splitext(os.path.basename(file))[0],
        "CAM_S_SF": data.get("CAM_S_SF"),
        "EEG_SessionID": data.get("EEG_SessionID"),
        "CAMSRecordedDTS": data.get("EvalDTS"),
        "EEGBeginDTS": data.get("SourceEEG_StartDTS")
    })
    print(f"\rProcessed {i}/{len(mat_files)}", end="")

df_CAMS = pd.DataFrame(rows)
df_CAMS["CAMSRecordedDTS"] = pd.to_datetime(df_CAMS["CAMSRecordedDTS"])
df_CAMS["EEGBeginDTS"] = pd.to_datetime(df_CAMS["EEGBeginDTS"])

print(f"\nTotal files processed: {len(df_CAMS)}")
print("Columns:", df_CAMS.columns.tolist())
```
